```python
sales = sales.withColumn(
    "net_sales",
    F.when(
        F.col("payment_id").isNull(),
        F.lit(0.0)
    ).otherwise(
        F.round(
            F.col("order_total") -
            F.coalesce(F.col("refund_amount"), F.lit(0.0)),
            2
        )
    )
)
```

```python
customers_clean = df.select(
    "customer_id",
    "country_code"
)
```

```python
sales = sales.join(
    customers_clean,
    "customer_id",
    "left"
)
```

```python
sales.count()
```

```python
sales_by_country = sales.groupBy(
    "country_code"
).agg(
    F.sum("net_sales").alias("country_revenue")
)
```

```python
sales_by_country = sales_by_country.withColumn(
    "country_revenue",
    F.round(F.col("country_revenue"), 2)
)
sales_by_country.show()
```

```python
total_sales = sales.agg(
    F.sum("net_sales").alias("total")
).first()["total"]

total_by_country = sales_by_country.agg(
    F.sum("country_revenue").alias("total")
).first()["total"]

print(f"Total de sales: {total_sales:.2f}")
print(f"Total por país: {total_by_country:.2f}")
print(f"Diferencia: {total_sales - total_by_country:.2f}")
```
